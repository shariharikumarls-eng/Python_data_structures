```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load Titanic dataset
df = pd.read_csv("Titanic-Dataset.csv")

# Plot using corrected dataframe columns
sns.swarmplot(
    data=df,
    x="Pclass",
    y="Age",
    hue="Sex",
    dodge=True,
    palette="Set1",
    size=4
)
plt.title("Age Distribution across Passenger Classes by Sex")
plt.xlabel("Passenger Class (Pclass)")
plt.ylabel("Age")
plt.legend(title="Sex")
plt.show()
```

```python
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt

# Upload file from your Windows PC
uploaded = files.upload()

# Load dataset
df = pd.read_csv("Titanic-Dataset.csv")

# Count male and female passengers
gender_counts = df['Sex'].value_counts()

# Data for pie chart
labels = gender_counts.index
sizes = gender_counts.values
explode = (0.1, 0)  # Highlight first slice (male)

# Create pie chart
plt.figure(figsize=(6, 6))
plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    explode=explode,
    shadow=True,
    startangle=90
)

plt.title('Percentage of Male vs Female Passengers')
plt.axis('equal')  # Makes the pie chart circular
plt.show()
```

```python
from google.colab import files
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Upload file from your Windows PC
uploaded = files.upload()

# Load dataset
df = pd.read_csv("Titanic-Dataset.csv")

# Select required columns with correct capitalization
corr_columns = ['Age', 'Fare', 'Pclass', 'SibSp', 'Parch', 'Survived']

# Compute correlation matrix
corr_matrix = df[corr_columns].corr()

# Create heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,      # Show correlation values
    cmap='coolwarm', # Color scheme
    fmt='.2f',       # Display 2 decimal places
    linewidths=0.5
)

plt.title('Correlation Matrix Heatmap')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load the dataset (df is already loaded in the kernel state, but we ensure it's ready)
if 'df' not in locals():
    df = pd.read_csv("Titanic-Dataset.csv")

# Create the violin plot
plt.figure(figsize=(10, 6))
sns.violinplot(
    data=df,
    x="Pclass",
    y="Fare",
    hue="Sex",
    split=True,
    inner="quart",
    palette="muted"
)

plt.title("Fare Distribution across Passenger Classes by Sex")
plt.xlabel("Passenger Class (Pclass)")
plt.ylabel("Fare")
plt.legend(title="Sex")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Ensure the dataset is loaded
if 'df' not in locals():
    df = pd.read_csv("Titanic-Dataset.csv")

# Create scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived",
    palette="Set1",
    alpha=0.8
)

plt.title("Relationship Between Age and Fare Colored by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived", labels=["No (0)", "Yes (1)"])
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
